# Q 23 How does the ratio of funded amount to loan amount vary across different loan statuses?

```python


df["Funding_Ratio"] = df["Funded Amount"] / df["Loan Amount"]

funding_summary = df.groupby("Loan Status")["Funding_Ratio"].agg(
    Mean="mean",
    Median="median",
    Min="min",
    Std="std",
    Count="count",
)
print("=== Funding Ratio Summary by Loan Status ===")
print((funding_summary * 100).round(2).astype(str) + "%")
```

# Q 24 Investigate the relationship between loan amount and total collection amount.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# 1. Load dataset
df = pd.read_csv('loan defaulter.csv')

# 2. Select relevant columns
col1 = 'Loan Amount'
col2 = 'Total Collection Amount'

# 3. Descriptive Statistics
print("=== Descriptive Statistics ===")
print(df[[col1, col2]].describe())

# 4. Correlation Analysis
pearson_r, pearson_p = stats.pearsonr(df[col1], df[col2])
spearman_r, spearman_p = stats.spearmanr(df[col1], df[col2])

print("\n=== Correlation Analysis ===")
print(f"Pearson Correlation (r)  : {pearson_r:.5f} (p-value: {pearson_p:.5f})")
print(f"Spearman Rank Correlation: {spearman_r:.5f} (p-value: {spearman_p:.5f})")

# 5. Scatter Plot Visualization
plt.figure(figsize=(10, 6))

# Plot scatter points with a linear regression line
sns.regplot(
    data=df,
    x=col1,
    y=col2,
    scatter_kws={'alpha': 0.3, 'color': '#1f77b4'},
    line_kws={'color': 'red', 'linewidth': 2}
)

# Formatting chart details
plt.title('Relationship Between Loan Amount and Total Collection Amount', fontsize=14, pad=15)
plt.xlabel('Loan Amount ($)', fontsize=12)
plt.ylabel('Total Collection Amount ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

# Annotate correlation on the plot
plt.text(
    0.05, 0.92,
    f'Pearson $r$: {pearson_r:.5f}\n$p$-value: {pearson_p:.3f}',
    transform=plt.gca().transAxes,
    fontsize=11,
    bbox=dict(boxstyle='round,pad=0.5', facecolor='white', alpha=0.8)
)

plt.tight_layout()
plt.show()
```
